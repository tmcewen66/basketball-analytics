```python
conn = sqlite3.connect('nba_stats.db')

query = '''
SELECT t.*, n.off_rating AS league_off_rating, n.ast_pct AS league_ast_pct, n.ts_pct AS league_ts_pct
FROM team_advanced_stats AS t
JOIN nba_advanced_averages AS n ON t.season_end_year = n.season_end_year
'''

team_league_advanced_df = pd.read_sql_query(query, conn)
conn.close()

team_league_advanced_df
```

```python
team_league_advanced_df['pts_plus'] = (team_league_advanced_df['off_rating'] / team_league_advanced_df['league_off_rating']) * 100
team_league_advanced_df['ts_plus'] = (team_league_advanced_df['ts_pct'] / team_league_advanced_df['league_ts_pct']) * 100
team_league_advanced_df['scoring_plus'] = 100 + (0.765 * (team_league_advanced_df["ts_plus"] - 100)) + (0.235 * (team_league_advanced_df["pts_plus"] - 100))

team_league_advanced_df[['team_name', 'season_end_year', 'pts_plus', 'ts_plus', 'scoring_plus']].head(10)
```

```python
scoring_plus_means = team_league_advanced_df.groupby('season_end_year')[['ts_plus', 'pts_plus', 'scoring_plus']].mean().reset_index()
scoring_plus_means
```
