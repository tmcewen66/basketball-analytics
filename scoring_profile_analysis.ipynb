```python
import sqlite3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
conn = sqlite3.connect('nba_stats.db')

query = '''
SELECT s.*, b.positions, b.made_field_goals, t.team_name, t.gp AS team_gp
FROM scoring_splits AS s
JOIN basic_stats AS b ON s.slug = b.slug AND s.season_end_year = b.season_end_year
JOIN team_per_100_stats AS t ON s.team_id = t.team_id AND s.season_end_year = t.season_end_year
'''
scoring_splits_df = pd.read_sql_query(query, conn)

conn.close()

pd.set_option('display.max_columns', None)
scoring_splits_df
```

```python
print(scoring_splits_df.groupby('positions')['pct_uast_fgm'].mean())
```

```python
plt.scatter(scoring_splits_df['made_field_goals'], scoring_splits_df['pct_uast_fgm'])
plt.xlabel('Made Field Goals')
plt.ylabel('Percentage of Unassisted Field Goals')

plt.show()
```

```python
scoring_splits_df['threshold'] = 300 * (scoring_splits_df['team_gp'] / 82)

scoring_splits_qual_df = scoring_splits_df[scoring_splits_df['made_field_goals'] >= scoring_splits_df['threshold']].copy()
scoring_splits_qual_df.reset_index(drop=True, inplace=True)

scoring_splits_qual_df
# The minimum made_field_goals threshold of 300 is the NBA's standard for qualifying for the FG% leaderboard.
```

```python
plt.scatter(scoring_splits_qual_df['made_field_goals'], scoring_splits_qual_df['pct_uast_fgm'])
plt.xlabel('Made Field Goals')
plt.ylabel('Percentage of Unassisted Field Goals')

plt.show()
```

```python
qual_means = scoring_splits_qual_df.groupby('positions')['pct_uast_fgm'].mean()
all_means = scoring_splits_df.groupby('positions')['pct_uast_fgm'].mean()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(qual_means.index, qual_means.values)
axes[0].set_xlabel('Positions')
axes[0].set_ylabel('Mean Percentage of Unassisted Field Goals')
axes[0].set_title('Mean % of Unassisted FG by Position (Qualified Players)')
axes[0].tick_params(axis='x', rotation=45)

axes[1].bar(all_means.index, all_means.values)
axes[1].set_xlabel('Positions')
axes[1].set_ylabel('Mean Percentage of Unassisted Field Goals')
axes[1].set_title('Mean % of Unassisted FG by Position (All Players)')
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
seasons_to_show = list(range(2001, 2027, 4))  # 2001, 2005, 2009, ...

fig, axes = plt.subplots(1, len(seasons_to_show), figsize=(4 * len(seasons_to_show), 5), sharey=True)

for ax, season in zip(axes, seasons_to_show):
    season_means = (
        scoring_splits_qual_df[scoring_splits_qual_df['season_end_year'] == season]
        .groupby('positions')['pct_uast_fgm']
        .mean()
    )
    ax.bar(season_means.index, season_means.values)
    ax.set_title(str(season))
    ax.set_xlabel('Positions')
    ax.tick_params(axis='x', rotation=45)

axes[0].set_ylabel('Mean Percentage of Unassisted Field Goals')
fig.suptitle('Mean % of Unassisted FG by Position, Every 4 Seasons (Qualified Players)')
plt.tight_layout()
plt.show()
```

```python
season_by_season_scoring_splits = scoring_splits_qual_df.groupby('season_end_year')['pct_uast_fgm'].mean().reset_index()
season_by_season_scoring_splits.rename(columns={'pct_uast_fgm': 'mean_pct_uast_fgm'}, inplace=True)
season_by_season_scoring_splits['num_players'] = scoring_splits_qual_df.groupby('season_end_year')['slug'].nunique().values
season_by_season_scoring_splits
```

```python
plt.bar(season_by_season_scoring_splits['season_end_year'], season_by_season_scoring_splits['mean_pct_uast_fgm'])
plt.xlabel('Season End Year')
plt.ylabel('Percentage of Unassisted Field Goals')
plt.ylim(0.3, 0.5)
plt.title('Percentage of Unassisted Field Goals by Season (Qualified Players)')
plt.show()
```

```python
players_by_season = scoring_splits_qual_df.groupby('season_end_year')['positions'].value_counts().unstack(fill_value=0)
players_by_season.plot(kind='bar', stacked=False, figsize=(12, 6))
plt.show()
```

```python
percentile_fractions = [i / 9 for i in range(1, 9)]  # 1/9, 2/9, ..., 8/9 -> 9 bins

pct_uast_fgm_cutoffs_df = (
    scoring_splits_qual_df
    .groupby('season_end_year')['pct_uast_fgm']
    .quantile(percentile_fractions)  # type: ignore[arg-type]
    .unstack()
)
pct_uast_fgm_cutoffs_df.columns = [f'{p * 100:.2f}_percentile' for p in percentile_fractions]
pct_uast_fgm_cutoffs_df = pct_uast_fgm_cutoffs_df.reset_index()
pct_uast_fgm_cutoffs_df
```

```python
cutoff_cols = [c for c in pct_uast_fgm_cutoffs_df.columns if c != 'season_end_year']

# Look up each player's season cutoffs, keeping row order aligned with scoring_splits_df
# so the comparisons below can be done position-wise instead of by index label.
season_cutoffs = scoring_splits_df[['season_end_year']].merge(
    pct_uast_fgm_cutoffs_df, on='season_end_year', how='left'
)

# A value exactly on a cutoff falls into the lower bin (bin edges are exclusive on the low end).
scoring_splits_df['uast_bin'] = 1 + sum(
    (scoring_splits_df['pct_uast_fgm'].values > season_cutoffs[col].values).astype(int)  # type: ignore
    for col in cutoff_cols
)

# Centers bin 5 (the middle bin) at 0, running from -4 (bin 1) to 4 (bin 9).
scoring_splits_df['uast_rating'] = scoring_splits_df['uast_bin'] - 5

scoring_splits_df['profile'] = np.select(
    [
        scoring_splits_df['uast_rating'] <= -2,
        scoring_splits_df['uast_rating'] <= 1,
    ],
    ['Finisher', 'Balanced'],
    default='Creator',
)

scoring_splits_df[['season_end_year', 'positions', 'pct_uast_fgm', 'uast_bin', 'uast_rating', 'profile']]
```
